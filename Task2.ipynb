```python
# ============================================================
# DATA ANALYSIS WITH PYTHON
# Dataset: sample-customers.csv
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")


# ============================================================
# 2. UPLOAD CSV FILE
# ============================================================

from google.colab import files

uploaded = files.upload()

print("\nFile uploaded successfully!")


# ============================================================
# 3. READ THE CSV DATASET
# ============================================================

# Read the uploaded CSV file
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("\nDataset loaded successfully!")


# ============================================================
# 4. DISPLAY FIRST 5 ROWS
# ============================================================

print("\n========== FIRST 5 ROWS ==========")
display(df.head())


# ============================================================
# 5. DISPLAY LAST 5 ROWS
# ============================================================

print("\n========== LAST 5 ROWS ==========")
display(df.tail())


# ============================================================
# 6. CHECK NUMBER OF ROWS AND COLUMNS
# ============================================================

print("\n========== DATASET SHAPE ==========")

rows, columns = df.shape

print("Number of Rows    :", rows)
print("Number of Columns :", columns)


# ============================================================
# 7. DISPLAY COLUMN NAMES
# ============================================================

print("\n========== COLUMN NAMES ==========")

print(df.columns.tolist())


# ============================================================
# 8. CHECK DATA TYPES
# ============================================================

print("\n========== DATA TYPES ==========")

print(df.dtypes)


# ============================================================
# 9. BASIC INFORMATION ABOUT DATASET
# ============================================================

print("\n========== DATASET INFORMATION ==========")

df.info()


# ============================================================
# 10. STATISTICAL SUMMARY
# ============================================================

print("\n========== STATISTICAL SUMMARY ==========")

display(df.describe(include="all"))


# ============================================================
# 11. CHECK MISSING VALUES
# ============================================================

print("\n========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

print(missing_values)


# ============================================================
# 12. CHECK MISSING VALUES IN PERCENTAGE
# ============================================================

print("\n========== MISSING VALUE PERCENTAGE ==========")

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)


# ============================================================
# 13. HANDLE MISSING VALUES
# ============================================================

print("\n========== HANDLING MISSING VALUES ==========")

# Fill missing values in text columns with "Unknown"
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].fillna("Unknown")

print("Missing values handled successfully!")


# ============================================================
# 14. CHECK DUPLICATE RECORDS
# ============================================================

print("\n========== DUPLICATE RECORDS ==========")

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)


# ============================================================
# 15. REMOVE DUPLICATE RECORDS
# ============================================================

df = df.drop_duplicates()

print("\nDuplicate records removed successfully!")


# ============================================================
# 16. CONVERT SIGNUP DATE INTO DATETIME
# ============================================================

print("\n========== DATE CONVERSION ==========")

df["signup_date"] = pd.to_datetime(df["signup_date"])

print(df["signup_date"].head())


# ============================================================
# 17. CREATE NEW DATE COLUMNS
# ============================================================

df["signup_year"] = df["signup_date"].dt.year
df["signup_month"] = df["signup_date"].dt.month
df["signup_month_name"] = df["signup_date"].dt.month_name()

print("\nNew date columns created successfully!")

display(df.head())


# ============================================================
# 18. CHECK UNIQUE VALUES
# ============================================================

print("\n========== UNIQUE VALUES ==========")

for column in ["city", "country", "segment"]:
    print("\n", column.upper())
    print(df[column].unique())


# ============================================================
# 19. COUNT CUSTOMERS BY SEGMENT
# ============================================================

print("\n========== CUSTOMERS BY SEGMENT ==========")

segment_count = df["segment"].value_counts()

print(segment_count)


# ============================================================
# 20. VISUALIZE CUSTOMERS BY SEGMENT
# ============================================================

plt.figure(figsize=(8, 5))

segment_count.plot(kind="bar")

plt.title("Number of Customers by Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()


# ============================================================
# 21. CUSTOMER PERCENTAGE BY SEGMENT
# ============================================================

print("\n========== CUSTOMER PERCENTAGE BY SEGMENT ==========")

segment_percentage = df["segment"].value_counts(normalize=True) * 100

print(segment_percentage.round(2))


# ============================================================
# 22. CUSTOMER COUNT BY CITY
# ============================================================

print("\n========== CUSTOMERS BY CITY ==========")

city_count = df["city"].value_counts()

print(city_count)


# ============================================================
# 23. TOP CITIES BY CUSTOMER COUNT
# ============================================================

print("\n========== TOP 10 CITIES ==========")

display(city_count.head(10))


# ============================================================
# 24. VISUALIZE CUSTOMERS BY CITY
# ============================================================

plt.figure(figsize=(10, 6))

city_count.plot(kind="bar")

plt.title("Number of Customers by City")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# 25. CUSTOMER COUNT BY COUNTRY
# ============================================================

print("\n========== CUSTOMERS BY COUNTRY ==========")

country_count = df["country"].value_counts()

print(country_count)


# ============================================================
# 26. VISUALIZE CUSTOMERS BY COUNTRY
# ============================================================

plt.figure(figsize=(10, 6))

country_count.plot(kind="bar")

plt.title("Number of Customers by Country")
plt.xlabel("Country")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# 27. CUSTOMER COUNT BY SIGNUP YEAR
# ============================================================

print("\n========== CUSTOMERS BY SIGNUP YEAR ==========")

year_count = df["signup_year"].value_counts().sort_index()

print(year_count)


# ============================================================
# 28. VISUALIZE SIGNUP YEAR
# ============================================================

plt.figure(figsize=(7, 5))

year_count.plot(kind="bar")

plt.title("Customer Signups by Year")
plt.xlabel("Year")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()


# ============================================================
# 29. CUSTOMER SIGNUPS BY MONTH
# ============================================================

print("\n========== CUSTOMER SIGNUPS BY MONTH ==========")

monthly_signup = df["signup_month_name"].value_counts()

print(monthly_signup)


# ============================================================
# 30. SORT MONTHS CORRECTLY
# ============================================================

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

monthly_signup = df["signup_month_name"].value_counts()

monthly_signup = monthly_signup.reindex(month_order).fillna(0)

print("\nMonthly signup count:")
print(monthly_signup)


# ============================================================
# 31. VISUALIZE MONTHLY SIGNUPS
# ============================================================

plt.figure(figsize=(12, 5))

monthly_signup.plot(kind="bar")

plt.title("Customer Signups by Month")
plt.xlabel("Month")
plt.ylabel("Number of Signups")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# 32. FIND MOST COMMON CUSTOMER SEGMENT
# ============================================================

most_common_segment = df["segment"].value_counts().idxmax()

print("\n========== MOST COMMON SEGMENT ==========")

print("Most common customer segment:", most_common_segment)


# ============================================================
# 33. FIND CITY WITH MOST CUSTOMERS
# ============================================================

most_common_city = df["city"].value_counts().idxmax()

print("\n========== CITY WITH MOST CUSTOMERS ==========")

print("City with most customers:", most_common_city)


# ============================================================
# 34. FIND COUNTRY WITH MOST CUSTOMERS
# ============================================================

most_common_country = df["country"].value_counts().idxmax()

print("\n========== COUNTRY WITH MOST CUSTOMERS ==========")

print("Country with most customers:", most_common_country)


# ============================================================
# 35. GROUP CUSTOMERS BY COUNTRY AND SEGMENT
# ============================================================

print("\n========== COUNTRY AND SEGMENT ANALYSIS ==========")

country_segment = pd.crosstab(
    df["country"],
    df["segment"]
)

display(country_segment)


# ============================================================
# 36. VISUALIZE COUNTRY VS SEGMENT
# ============================================================

country_segment.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Customer Segments by Country")
plt.xlabel("Country")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)

plt.show()


# ============================================================
# 37. GROUP CUSTOMERS BY CITY AND SEGMENT
# ============================================================

print("\n========== CITY AND SEGMENT ANALYSIS ==========")

city_segment = pd.crosstab(
    df["city"],
    df["segment"]
)

display(city_segment)


# ============================================================
# 38. CHECK EMAIL DUPLICATES
# ============================================================

print("\n========== DUPLICATE EMAILS ==========")

duplicate_emails = df[df["email"].duplicated(keep=False)]

print("Number of records with duplicate emails:",
      len(duplicate_emails))

display(duplicate_emails.head(20))


# ============================================================
# 39. CHECK DUPLICATE CUSTOMER IDs
# ============================================================

print("\n========== DUPLICATE CUSTOMER IDs ==========")

duplicate_ids = df["customer_id"].duplicated().sum()

print("Duplicate customer IDs:", duplicate_ids)


# ============================================================
# 40. SORT CUSTOMERS BY SIGNUP DATE
# ============================================================

print("\n========== LATEST CUSTOMER SIGNUPS ==========")

latest_customers = df.sort_values(
    by="signup_date",
    ascending=False
)

display(
    latest_customers[
        ["customer_id", "name", "city", "country", "signup_date", "segment"]
    ].head(10)
)


# ============================================================
# 41. EARLIEST CUSTOMER SIGNUPS
# ============================================================

print("\n========== EARLIEST CUSTOMER SIGNUPS ==========")

earliest_customers = df.sort_values(
    by="signup_date",
    ascending=True
)

display(
    earliest_customers[
        ["customer_id", "name", "city", "country", "signup_date", "segment"]
    ].head(10)
)


# ============================================================
# 42. FINAL CLEANED DATASET
# ============================================================

print("\n========== CLEANED DATASET ==========")

display(df.head())


# ============================================================
# 43. FINAL DATASET INFORMATION
# ============================================================

print("\n========== FINAL DATASET INFORMATION ==========")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 44. FINAL MISSING VALUE CHECK
# ============================================================

print("\n========== FINAL MISSING VALUE CHECK ==========")

print(df.isnull().sum())


# ============================================================
# 45. FINAL DUPLICATE CHECK
# ============================================================

print("\n========== FINAL DUPLICATE CHECK ==========")

print("Duplicate rows:", df.duplicated().sum())


# ============================================================
# 46. IMPORTANT DATASET SUMMARY
# ============================================================

print("\n====================================================")
print("              DATA ANALYSIS SUMMARY")
print("====================================================")

print("Total Customers :", len(df))
print("Total Columns   :", len(df.columns))

print("Most Common Segment :", most_common_segment)
print("Most Common City    :", most_common_city)
print("Most Common Country :", most_common_country)

print("\nCustomer Segments:")
print(df["segment"].value_counts())

print("\nTop 5 Cities:")
print(df["city"].value_counts().head(5))

print("\nTop 5 Countries:")
print(df["country"].value_counts().head(5))


# ============================================================
# 47. SAVE CLEANED DATASET
# ============================================================

output_file = "cleaned_sample_customers.csv"

df.to_csv(output_file, index=False)

print("\nCleaned dataset saved as:", output_file)


# ============================================================
# 48. DOWNLOAD CLEANED DATASET
# ============================================================

from google.colab import files

files.download(output_file)


# ============================================================
# 49. FINAL CONCLUSION
# ============================================================

print("""
====================================================
                 FINAL CONCLUSION
====================================================

1. The customer dataset was successfully imported using Pandas.

2. The dataset was explored using head(), tail(), shape,
   columns, info() and describe().

3. Missing values were checked and handled.

4. Duplicate records were identified and removed.

5. The signup_date column was converted into datetime format.

6. Customer data was analyzed based on city, country and
   customer segment.

7. Different visualizations were created using Matplotlib.

8. Customer signup trends were analyzed by year and month.

9. The cleaned dataset was saved as a new CSV file.

10. Pandas and NumPy were used for data analysis and
    data manipulation.
====================================================
""")

print("DATA ANALYSIS PROJECT COMPLETED SUCCESSFULLY!")
```
